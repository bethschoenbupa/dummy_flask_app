```python
import requests
import time
from pprint import pprint
```

# API v1 Demo
Beneficial for stakeholders to view the an example output from the app without having to disect the code/run the app themselves.

```python
name = "Vasiliki"
birthday = "2025-12-19"

body = {
    "name":name,
    "date":birthday
}

url = f'http://localhost:5000/api/v1/birthdayhaiku'
start = time.time()
x = requests.post(url, json=body)
end = time.time()

time_elapsed = end-start
print(f"Time elapsed: {time_elapsed}s")

result = x.json()

pprint(result)
```
