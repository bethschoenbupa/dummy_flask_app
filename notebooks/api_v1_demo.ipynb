```python
import requests
import time
from pprint import pprint
from datetime import datetime, timedelta
```

# API v1 Demo
Beneficial for stakeholders to view an example output from the app without having to disect the code/run the app themselves. They may also be interested in viewing runtimes too. 

## Birthday Haiku
Given a person and date, determine whether it's the person's birthday. Write a Haiku depending on whether it's their birthday?

#### `isBirthday = True`

```python
name = "Vasiliki"
today = datetime.today()
birthday =today.strftime("%Y-%m-%d")

body = {
    "name":name,
    "date":birthday
}

url = f'http://localhost:5000/api/v1/birthdayhaiku'
start = time.time()
x = requests.post(url, json=body)
end = time.time()

time_elapsed = end-start
print(f"Time elapsed: {time_elapsed}s")

result = x.json()

pprint(result)
```

#### `isBirthday = False`

```python
name = "Vasiliki"
yesterday = datetime.today() - timedelta(days=1)
birthday = yesterday.strftime("%Y-%m-%d")

body = {
    "name":name,
    "date":birthday
}

url = f'http://localhost:5000/api/v1/birthdayhaiku'
start = time.time()
x = requests.post(url, json=body)
end = time.time()

time_elapsed = end-start
print(f"Time elapsed: {time_elapsed}s")

result = x.json()

pprint(result)
```
