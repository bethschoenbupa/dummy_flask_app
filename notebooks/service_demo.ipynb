# Service Demo
If we want to explore endpoint functionality without running the app/worrying about app infrastructure, we can run services in the notebook. We need to ensure the generator factory is setup, allowing us to specify provider and shadow mode booleans. 

```python
today = datetime.today()
birthday = today.strftime("%Y-%m-%d")

INPUT_DATA = {
    "name": "Beth",
    "date":birthday
}

service_input = HaikuRequestModel(**INPUT_DATA)
# this isn't needed if you're using the prompt version(s) specified in prompt_versions.py
testing_prompt_versions = {
    vr.haiku_prompt_name: 1,
}
```

```python
generator_factory = TextGeneratorFactory(ai_context={"provider":"vertex", "shadow_mode":True}, model_map=MODEL_MAP)

result = haiku_service(service_input, generator_factory.get_generator, prompt_map=testing_prompt_versions)

print(json.dumps(result.model_dump(by_alias=False), indent=2))
```

```python
generator_factory = TextGeneratorFactory(ai_context={"provider":"vertex", "shadow_mode":False}, model_map=MODEL_MAP)

result = haiku_service(service_input, generator_factory.get_generator, prompt_map=testing_prompt_versions)

print(json.dumps(result.model_dump(by_alias=False), indent=2))
```
